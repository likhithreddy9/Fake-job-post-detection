```python
vectorizer = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)
```

```python
X = vectorizer.fit_transform(df["text"])
```

```python
df["text_length"] = df["text"].str.len()
```

```python
print(df[["text", "text_length"]].head())
```

```python
df["exclamation_count"] = df["text"].str.count("!")
print(
    df[["text", "exclamation_count"]].head()
)
```

```python
df["has_salary"] = (
    df["text"]
    .str.contains(
        r"\$|salary|pay|wage|compensation",
        case=False,
        regex=True
    )
    .astype(int)
)
```

```python
print(
    df[
        [
            "text_length",
            "exclamation_count",
            "has_salary"
        ]
    ].head()
)
```
